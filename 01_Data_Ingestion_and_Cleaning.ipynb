```python
def to_datestamp (df: DataFrame) -> DataFrame:
    df_transformed = df \
        .withColumn("order_date__DateOrders_", f.to_date(f.col("order_date__DateOrders_"), "D-M-yyyy HH:mm")) \
        .withColumn("shipping_date__DateOrders_", f.to_date(f.col("shipping_date__DateOrders_"), "D-M-yyyy HH:mm"))
    decim_columns = ["Benefit_per_order", "Sales_per_customer", "Latitude", "Longitude", "Order_Item_Discount", "Order_Item_Discount_Rate", "Order_Item_Profit_Ratio", "Order_Item_Product_Price", "Order_Item_Total", "Order_Profit_Per_Order"]
    for col in decim_columns:
        if col in df_transformed.columns:
            df_transformed = df_transformed.withColumn(col, f.col(col).cast("double")
            )
    return df_transformed
```

```python
df1_raw = spark.table("table_1")
df_typed = to_datestamp(df1_raw) 
```

```python
def missing_shippements(df: DataFrame) -> DataFrame:
    shipping_col = "shipping_date__DateOrders_"
    if shipping_col in df.columns:
        df_transform = df.withColumn(
                "shippement_status", 
                F.when(F.col(shipping_col).isNull(), F.lit("UNFULFILLED_OR_PENDING"))
                .otherwise(F.lit("Shipped"))
        )
        return df_transform
    return df
```

```python
from pyspark.sql import functions as F
df1 = spark.table("table_1")
df1_cleaned = missing_shippements(df1)
```

```python
from pyspark.sql import DataFrame
from pyspark.sql import functions as F

def standardize_countries(df: DataFrame) -> DataFrame:
    col_name = "Order_Country"
    if col_name not in df.columns:
        return df
    
    # Map Spanish translations, stripped characters, and original corrupted variants
    country_map = {
        # Stripped & Corrupted Patterns
        "Repblica Dominicana": "Dominican Republic",
        "Repblica Democrtica del Congo": "Democratic Republic of the Congo",
        "Per": "Peru",
        "Kirguistn": "Kyrgyzstan",
        "Barin": "Bahrain",
        "Kazajistn": "Kazakhstan",
        "SudAfrica": "South Africa",
        "Mxico": "Mexico",
        "Hait": "Haiti",
        "Pases Bajos": "Netherlands",
        "Espaa": "Spain",
        "Turqua": "Turkey",
        "Japn": "Japan",
        
        # Standard Spanish Translations
        "Irak": "Iraq",
        "Kenia": "Kenya",
        "Rusia": "Russia",
        "Suiza": "Switzerland",
        "Siria": "Syria",
        "Marruecos": "Morocco",
        "Dinamarca": "Denmark",
        "Costa de Marfil": "Ivory Coast",
        "Trinidad y Tobago": "Trinidad and Tobago",
        "Brasil": "Brazil",
        "Estados Unidos": "United States",
        "Reino Unido": "United Kingdom",
        "Alemania": "Germany",
        "Francia": "France",
        "Polonia": "Poland",
        "Ruanda": "Rwanda",
    }
    
    # 1. Strip non-ASCII characters
    df_clean = df.withColumn(
        col_name, 
        F.regexp_replace(F.trim(F.col(col_name)), r"[^\x00-\x7F]+", "")
    )
    
    # 2. Apply mapping
    mapping_expr = F.create_map([F.lit(x) for chain in country_map.items() for x in chain])
    
    return df_clean.withColumn(
        col_name, 
        F.coalesce(mapping_expr[F.col(col_name)], F.col(col_name))
    )

# Test on Table 1
df1_clean = standardize_countries(df1_cleaned)
df1_clean.select("Order_Country").distinct().show(truncate=False)
```
