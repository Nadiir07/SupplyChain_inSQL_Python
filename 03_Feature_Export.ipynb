```python
import pandas as pd

# Limit to 10,000 transaction rows to stay well within memory limits
sampled_df = spark.sql("""
    SELECT 
        s.late_delivery_risk AS label,
        l.market,
        l.order_region,
        s.shipping_mode,
        p.category_name,
        MONTH(s.shipping_date) AS order_month,
        DAYOFWEEK(s.shipping_date) AS order_day_of_week
    FROM fact_shipments s
    JOIN fact_orders o ON s.order_item_id = o.order_item_id
    JOIN dim_locations l ON o.location_id = l.location_id
    JOIN dim_products p ON o.product_id = p.product_id
    LIMIT 10000
""").dropna().toPandas()

sampled_df.to_csv("model_features_sampled.csv", index=False)
print("Sampled CSV saved successfully without crashing the kernel!")
```
